```python
import pandas as pd
import nltk
import string

full_df = pd.read_csv('tweet_small.csv')
```

```python
print(full_df.head())
```

```python
df = full_df[['text']]
df.info()
```

```python
print(df)
```

```python
df['text_lower'] = df['text'].str.lower()
print(df.head())
```

```python
PUNCT_TO_REMOVE = string.punctuation
def remove_punct(text):
    return text.translate(str.maketrans('', '', PUNCT_TO_REMOVE))

df['text_wo_punct'] = df['text_lower'].apply(remove_punct)
print(df.head())
```

```python
df.head()
```

```python
PUNCT_TO_REMOVE
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
', '.join(stopwords.words('english'))
```

```python
STOPWORDS = set(stopwords.words('english'))
def remove_stopwords(text):
    return ' '.join([word for word in str(text).split(' ') if word not in STOPWORDS])

df['text_wo_stop'] = df['text_wo_punct'].apply(remove_stopwords)
df.head()
```

```python
from collections import Counter
cnt = Counter()
for text in df['text_wo_stop'].values:
    for word in text.split():
        cnt[word] += 1

cnt.most_common(10)
```

```python
FREQWORDS = set([w for (w, _) in cnt.most_common(10)])

def remove_freqwords(text):
    return ' '.join([word for word in str(text).split(' ') if word not in FREQWORDS])

df['text_wo_stop_freq'] = df['text_wo_stop'].apply(remove_freqwords)
df.head()
```

```python
n_rare_words = 10
RAREWORDS = set([w for (w, _) in cnt.most_common()[:-n_rare_words-1:-1]])
RAREWORDS
```

```python
def remove_rarewords(text):
    return ' '.join([word for word in str(text).split(' ') if word not in RAREWORDS])

df['text_wo_stop_freq_rare'] = df['text_wo_stop_freq'].apply(remove_rarewords)
df.head()
```

```python
from nltk.stem.porter import PorterStemmer
stemmer = PorterStemmer()

def stem_words(text):
    return ' '.join([stemmer.stem(word) for word in str(text).split(' ')])

df['text_stemmed'] = df['text_wo_stop_freq_rare'].apply(stem_words)
df.head()

```

```python
from nltk.stem.snowball import SnowballStemmer
stemmer = SnowballStemmer('russian')
wrds = ['летел','слил','наложил','убегал','бежать','бегать', 'приходить', 'трансцендентный', 'урегулирование', 'пишущий']
for w in wrds:
    print(w, stemmer.stem(w))
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemmatize = WordNetLemmatizer()
wrds = ['running', 'runs', 'renderer', 'predefine', 'paranormal']
for w in wrds:
    print(w, lemmatize.lemmatize(w, 'n'))
```
